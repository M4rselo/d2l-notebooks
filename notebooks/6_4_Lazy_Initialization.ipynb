## Lazy Initialization

```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
net = nn.Sequential(nn.LazyLinear(256), nn.ReLU(), nn.LazyLinear(10))
```

```python
net[0].weight
```

```python
X = torch.rand(2, 20)
net(X)

net[0].weight.shape
```

```python
net[0].weight
```

```python
@d2l.add_to_class(d2l.Module)  #@save
def apply_init(self, inputs, init=None):
    self.forward(*inputs)
    if init is not None:
        self.net.apply(init)
```
