# Padding and Stride

```python
import torch
from torch import nn
```

## Padding

```python
def comp_conv2d(conv2d, X):
    X = X.reshape((1, 1) + X.shape)
    Y = conv2d(X)
    return Y.reshape(Y.shape[2:])

conv2d = nn.LazyConv2d(1, kernel_size=3, padding=1)
X = torch.rand(size=(8, 8))
comp_conv2d(conv2d, X).shape
```

```python
conv2d = nn.LazyConv2d(1, kernel_size=(5, 3), padding=(2, 1))
comp_conv2d(conv2d, X).shape
```

```python
comp_conv2d(conv2d, X)
```

### Ex. 1

```python
def calc_dims(n_dims, k_dims, p_dims=(0, 0), s_dims=(1, 1)):
    dim1 = (n_dims[0]-k_dims[0]+p_dims[0]+s_dims[0])//s_dims[0]
    dim2 = (n_dims[1]-k_dims[1]+p_dims[1]+s_dims[1])//s_dims[1]
    return f"Output size: {dim1} x {dim2}"
```

```python
print(comp_conv2d(conv2d, X).shape)
calc_dims((8, 8), (3, 5), (0, 1), (3, 4))
```
